```python
#16.3 Create a CSV file called books2.csv by using these lines:
#title,author,year
#The Weirdstone of Brisingamen,Alan Garner,1960
#Perdido Street Station,China Miéville,2000
#Thud!,Terry Pratchett,2005
#The Spellman Files,Lisa Lutz,2007
#Small Gods,Terry Pratchett,1992
```

```python
import csv
books2 = [
    ["The Weirdstone of Brisingamen", "Alan Garner", 1960],
    ["Perdido Street Station", "China Miéville", 2000],
    ["Thud!", "Terry Pratchett", 2005],
    ["The Spellman Files", "Lisa Lutz", 2007],
    ["Small Gods", "Terry Pratchett", 1992],
    ]
with open("books2.csv", "wt") as fout:
    csvout = csv.writer(fout)
    csvout.writerows(books2)
```

```python
#16.4 Use the sqlite3 module to create a SQLite database called books.db and 
#a table called books with these fields: title (text), author (text), and year (integer).
```

```python
#16.5 Read books2.csv and insert its data into the book table.
```

```python
import sqlite3
conn = sqlite3.connect('books.db')
curs = conn.cursor()
curs.execute('''CREATE TABLE books
    (title TEXT,
    author TEXT,
    year INT)''')
```

```python
import csv
with open("books2.csv", "rt") as fin:
    cin = csv.reader(fin)
    for book in cin:
        curs.execute('''INSERT INTO books (title, author, year) VALUES (?, ?, ?)''',
            (book[0], book[1], book[2]))
conn.commit()
```

```python
#16.8 Use the sqlalchemy module to connect to the sqlite3 database books.db that you just made in exercise 16.4. 
#As in 16.6, select and print the title column from the book table in alphabetical order.
```

```python
import sqlalchemy as sa
engine = sa.create_engine("sqlite:///books.db")
with engine.connect() as conn:
    result = conn.execute(sa.text("SELECT title FROM books ORDER BY title"))
    for row in result:
        print(row[0])
```
